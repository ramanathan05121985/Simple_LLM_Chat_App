ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt=ChatPromptTemplate.from_messages([
    ('system','You are a professional translator.Translate the following text from English to {target_language}'),
    ('human','{text}')
    ])
```

```python
chain=prompt|llm
```

```python
msg_trans=chain.invoke({"target_language":"Tamil","text":"What is the capital of India"})
```

```python
msg_trans.content
```
